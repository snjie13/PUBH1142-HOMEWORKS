## Quiz Question

```python
import numpy
from scipy import stats

# Create population
numpy.random.seed(28)
dBP = numpy.random.normal(85, 10, 100000)

# Use seed 12
numpy.random.seed(12)
```

### Question 1

```python
numpy.random.seed(12)
sample = numpy.random.choice(dBP, 40)
```

### Question 2

```python
# Sample mean
sample_mean = sample.mean()
print("Sample Mean:", sample_mean)
```

### Question 3

```python
# Sample standard deviation
sample_std = sample.std(ddof=1)
print("Sample Std:", sample_std)
```

### Question 4

```python
# Standard error
standard_error = sample_std / numpy.sqrt(40)
print("Standard Error:", standard_error)
```

### Question 5

```python
# Confidence coefficient
c = stats.t.ppf(0.975, 39)
print("Confidence Coefficient:", c)
```

### Question 6

```python
# Confidence interval
lower_bound = sample_mean - c * standard_error

print("Lower Bound:", lower_bound)
```

### Question 7

```python
upper_bound = sample_mean + c * standard_error
print( "upper bound:", upper_bound)
```

### Question 8

```python
bootstrapped_means = numpy.array([
    numpy.random.choice(sample, 40, replace=True).mean()
    for _ in range(1000)
])

bootstrapped_means.sort()
```

### Question 9

```python
#  2.5th percentile
print(bootstrapped_means[24])
```
